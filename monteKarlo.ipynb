```python

#10D sphere volume
import numpy as np
from numpy import random
import matplotlib.pyplot as plt

D=10
#number of dimensions
N=5#number of points in each dimension
#generate uniform grid from -1 to 1,
xarr=np.linspace(-1,1,N)

#total number of points
TN=np.power(N,D)

count = 0
for i1 in range (0,N):
  x1 = xarr[i1]   
  for i2 in range (0,N):
    x2 = xarr[i2]  
    for i3 in range (0,N):
      x3 = xarr[i3]
      for i4 in range (0,N):
        x4 = xarr[i4]
        for i5 in range (0,N):
          x5 = xarr[i5]
          for i6 in range (0,N):
            x6 = xarr[i6]
            for i7 in range (0,N):
              x7 = xarr[i7]
              for i8 in range (0,N):
                x8 = xarr[i8]
                for i9 in range (0,N):
                  x9 = xarr[i9]
                  for i10 in range (0,N):
                    x10 = xarr[i10]
                    rsq=(x1**2+x2**2+x3**2+x4**2+x5**2+x6**2+x7**2+x8**2+x9**2+x10**2)
              
                    if (rsq<=1.0):
                      count+=1
                  
Volume = (count/TN)*(2**10)
                        
print("volume=",Volume) 
```

```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt

N = 1000  # Set number of points

def func(x):
    return np.cos(x)

# Define integration bounds
a = 0
b = 3 * np.pi / 2

# Generate random x and y points
xpts = random.rand(N) * (b - a) + a       # x in [0, 3π/2]
ypts = random.rand(N) * 2 - 1.0           # y in [-1, 1]

positive = 0
negative = 0

for i in range(N):
    yval = func(xpts[i])
    plt.scatter(xpts[i], yval, color='red', s=5)   # Actual function value (red dots)
    plt.scatter(xpts[i], ypts[i], color='blue', s=5)  # Sampled point (blue dots)
    
    if yval > 0 and 0 < ypts[i] <= yval:
        positive += 1
    elif yval < 0 and 0 > ypts[i] >= yval:
        negative += 1

# Monte Carlo integration
integral = (b - a) * (positive - negative) / N
print("Estimated integral:", integral)

# Plot function curve
x = np.linspace(a, b, 500)
plt.plot(x, func(x), color='black', label='cos(x)')

plt.axhline(0, color='gray', linestyle='--')
plt.title("Monte Carlo Integration of cos(x) on [0, 3π/2]")
plt.legend()
plt.show()
```

```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt

x=np.array([0.5])
dx=0.1

def target(y):
    if (y<1.0E-16 or y>1.0):
        tg = 0.0
    else:
        tg = -np.log(y)*np.sin(y)
    return tg

vtarget = np.vectorize(target)

N=1000

xlin = np.linspace(0,1,100)
tar = vtarget(xlin)
norm=np.sum(tar)*0.01
plt.plot(xlin,tar/norm)

for i in range (0,N):
    cstep = x[-1]
    step = dx*(random.rand()-0.5)
    nstep = cstep+step
    #if(nstep<0 or nstep >1.0):
    #    continue
    accept = min(1.0,target(nstep)/target(cstep))
    if(random.rand()<accept):
        x=np.append(x,nstep)
    else:
        x=np.append(x,cstep)
    

    
counts,bins = np.histogram(x,bins=10)
deltax = bins[1]-bins[0]
bin_mid = (bins[1:]+bins[:-1])/2.0
plt.plot(bin_mid,(counts/np.size(x))/deltax)
plt.show()
        
```

```python

import numpy as np
from numpy import random
import matplotlib.pyplot as plt

N=10000

x1 = np.random.rand(N)
x2 = np.random.rand(N)

r1 = np.sqrt(-2.0*np.log(x1))
theta = x2*2*np.pi

y1 = r1*np.cos(theta)
y2 = r1*np.sin(theta)

plt.hist(y1,bins=30)
plt.show()

counts,bins = np.histogram(y1,bins=50)
deltax = bins[1]-bins[0]

pdf = (counts/N)/deltax

bin_mid = (bins[0:-1]+bins[1:])/2

#Gaussian
gaussian = (1.0/np.sqrt(2*np.pi))*np.exp(-0.5*bin_mid**2)

plt.plot(bin_mid,pdf)
plt.plot(bin_mid,gaussian)
plt.yscale("log")
plt.show()
```
