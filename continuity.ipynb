```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x0=0
xf=1
N=10
delta_x=(xf-x0)/N


t0=0
tf=10


delta_t=0.001
u=0.1

```

```python
x=np.linspace(0,1,num=N+2)
p=np.exp(-((x-0.5)/0.15)**2)
a=u*delta_t/2*delta_x 
```

```python
T=np.zeros((N,N))
for i in range(N):
    T[i][i]=1
    if(i>0):
        T[i][i-1]=a
    if(i<N-1):
        T[i][i+1]=-a
```

```python
t=t0
i=0
while (t<tf):
    p[1:N+1]=p[1:N+1]-a* (np.matmul(T,p[1:N+1]))
    t=t+delta_t
    if(t >i):
        i+=2
        plt.figure()
        plt.plot(x,p)
    
```

```python
T=np.zeros((N,N))
for i in range(N):
    T[i][i]=1
    if(i>0):
        T[i][i-1]=-a
    if(i<N-1):
        T[i][i+1]=a
```

```python
t=t0
i=0
T=np.linalg.inv(T)
while (t<tf):
    p[1:N+1]=p[1:N+1]-a* (np.matmul(T,p[1:N+1]))
    t=t+delta_t
    if(t >i):
        i+=2
        plt.figure()
        plt.plot(x,p)
    
```
